```python
import pandas as pd
import random
import numpy as np
from faker import Faker

fake = Faker()

# Constants
courses = ['BSc-IT', 'BSc-DS']
years = ['FY', 'SY', 'TY']
semesters = [1, 2]
gender_options = ['Male', 'Female', 'Other']
caste_options = ['General', 'OBC', 'SC', 'ST']
enrollment_status_options = ['Active', 'Dropped Out', 'Graduated']

# Helper to generate student IDs
def generate_student_id(course, year, index):
    prefix = {'FY': '25', 'SY': '24', 'TY': '23'}
    course_code = 'IT' if course == 'BSc-IT' else 'DS'
    return f"{prefix[year]}{course_code}{index:02d}"

# Create a single student record
def create_student(course, year, index):
    semester = random.choice(semesters)
    student_id = generate_student_id(course, year, index)
    name = fake.name()
    gender = random.choice(gender_options)
    age = random.randint(18, 23)
    caste = random.choice(caste_options)
    income = random.randint(100000, 1000000)
    internal = random.randint(20, 30)
    theory = random.randint(30, 70)
    practical = random.randint(20, 50)
    backlogs = random.randint(0, 3)
    fees_paid = random.randint(20000, 50000)
    fees_date = fake.date_this_year()
    scholarship = random.choice(['Yes', 'No'])
    teacher_rating = round(random.uniform(3.0, 5.0), 1)
    teacher_feedback = fake.sentence(nb_words=6)
    library_usage = random.randint(0, 50)
    medical_history = random.choice(['Yes', 'No'])
    discipline = random.choice(['Yes', 'No'])
    academic_year = f"202{random.randint(2,4)}-202{random.randint(5,6)}"
    attendance = round(random.uniform(50.0, 100.0), 2)
    assignment_rate = round(random.uniform(60.0, 100.0), 2)
    class_participation = round(random.uniform(1.0, 5.0), 2)
    late_submissions = random.randint(0, 5)
    lms_login = random.randint(10, 200)
    videos_watched = random.randint(5, 50)
    forum_posts = random.randint(0, 20)
    study_hours = round(random.uniform(1.0, 10.0), 1)
    assign_time_avg = round(random.uniform(1.0, 5.0), 1)
    parent_edu = random.choice(['None', 'School', 'UG', 'PG'])
    net_home = random.choice(['Yes', 'No'])
    stress = random.choice(['Low', 'Medium', 'High'])
    mental_score = random.randint(1, 10)
    counselling = random.choice(['Yes', 'No'])
    clubs = random.randint(0, 3)
    events = random.randint(0, 5)
    volunteer = random.choice(['Yes', 'No'])
    enrollment_status = random.choice(enrollment_status_options)

    return [
        student_id, name, course, year, semester, gender, age, caste, income,
        internal, theory, practical, backlogs, fees_paid, fees_date, scholarship,
        teacher_rating, teacher_feedback, library_usage, medical_history, discipline,
        academic_year, attendance, assignment_rate, class_participation,
        late_submissions, lms_login, videos_watched, forum_posts, study_hours,
        assign_time_avg, parent_edu, net_home, stress, mental_score, counselling,
        clubs, events, volunteer, enrollment_status
    ]

# Generate dataset
columns = [
    "Student ID", "Name", "Course", "Year", "Semester", "Gender", "Age", "Caste",
    "Family Income", "Internal Marks", "Theory Marks", "Practical Marks", "Backlogs",
    "Fees Paid", "Fees Payment Date", "Scholarship Received", "Teacher Rating",
    "Teacher Feedback", "Library Usage", "Medical History Reported",
    "Disciplinary Action", "Academic Year", "Attendance %", "Assignment Submission Rate",
    "Class Participation", "Late Submissions", "LMS Login Count",
    "Video Lectures Watched", "Forum Participation", "Study Hours/Week",
    "Assignment Completion Time (Avg)", "Parental Education",
    "Internet Availability at Home", "Stress Level", "Mental Health Score",
    "Counselling Opted", "Clubs Joined", "Events Participation", "Volunteer Activities",
    "Enrollment Status"
]

data = []
for course in courses:
    for year in years:
        for i in range(1, 201):
            student = create_student(course, year, i)
            data.append(student)

df = pd.DataFrame(data, columns=columns)
csv_path = "D:/path_finder/fake_student_performance_dataset.csv"
df.to_csv(csv_path, index=False)
csv_path
```
